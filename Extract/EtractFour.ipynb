```python
import SRC
import os
from SRC.helpFolder import make_weather_data_dataframe, get_weather_data_for_city, append_to_json_file

metadata = SRC.load_model_metadata()
```

```python
metadata['weatherInformation']
```

```python
weather_data = get_weather_data_for_city(
    start_date=metadata['PMSInformation']['start_date_range'],
    end_date=metadata['weatherInformation']['weather_end_date'],
    city=metadata['weatherInformation']['default_city']
)

weather_data
```

```python
append_to_json_file(
    weather_data,
    '../Data/WeatherData/weather_data.json',
)
```

```python
import json

with open('../Data/WeatherData/weather_data.json', 'w', encoding='utf-8') as f:
    json.dump(weather_data, f, indent=4, ensure_ascii=False)
```
