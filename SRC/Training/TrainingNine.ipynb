```python
def transform_to_dataframe(data):
    """
    Transforms the input data dictionary into a Pandas DataFrame.

    Args:
        data (dict): A dictionary structured with dates as keys, 
                     and nested dictionaries for segments and their metrics.
    """
    records = []
    for date, segments_data in data.items():
        for segment, metrics in segments_data.items():
            records.append({
                'date': date,
                'segment': segment,
                'taux_occupation': metrics.get('Occupancy percentage'),
                'total_price': metrics.get('Revenue')
            })
    df = pd.DataFrame(records)
    return df
```
